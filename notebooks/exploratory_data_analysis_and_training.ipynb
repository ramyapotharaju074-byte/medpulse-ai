**Project Goal:** Build an end-to-end clinical predictive health engine for cardiac risk forecasting, model evaluation, Explainable AI (XAI), and MLOps drift monitoring.

```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set project root
project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(project_root))

from src import config
from src.data.data_generator import generate_synthetic_medical_data
from src.data.preprocessing import prepare_data
from src.models.model_trainer import train_and_compare_models
from src.models.evaluator import generate_evaluation_report
from src.explainability.explainability import get_global_feature_importance, explain_patient_risk
from src.monitoring.drift_monitor import analyze_dataset_drift
```

## 1. Data Ingestion & Exploratory Data Analysis (EDA)

```python
# Generate benchmark medical dataset
df = generate_synthetic_medical_data(num_samples=2500, random_seed=42)
print(f"Dataset Shape: {df.shape}")
    df.head()
```

```python
# Inspect class distribution & missing value summary
print("--- Target Risk Distribution ---")
print(df['cardiac_risk'].value_counts(normalize=True))
print("\n--- Missing Values Summary ---")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

## 2. Preprocessing & Feature Engineering Pipeline

```python
data_dict = prepare_data(df, test_size=0.20)
print(f"X_train Transformed Shape: {data_dict['X_train'].shape}")
print(f"X_test Transformed Shape: {data_dict['X_test'].shape}")
```

## 3. Multi-Model Training & 5-Fold Stratified Cross Validation

```python
train_results = train_and_compare_models(data_dict)
champion_name = train_results['best_model_name']
champion_model = train_results['trained_models'][champion_name]
```

## 4. Evaluation & Performance Benchmarking

```python
eval_report = generate_evaluation_report(data_dict['y_test'], train_results['results'])
pd.DataFrame(eval_report['summary_table'])
```

## 5. Explainable AI (XAI) — Global Importance & Local SHAP Waterfall

```python
global_imp = get_global_feature_importance(champion_model, data_dict['feature_names'])
pd.DataFrame(global_imp)
```

```python
# Explain individual patient risk
sample_patient = data_dict['X_test_raw'].iloc[[0]]
explanation = explain_patient_risk(sample_patient, data_dict['preprocessor'], champion_model, data_dict['feature_names'])
print(f"Risk Category: {explanation['risk_category']} ({explanation['risk_percentage']}%")
pd.DataFrame(explanation['top_drivers'])
```

## 6. MLOps Data Drift Analysis (PSI & KS Test)

```python
# Simulate shift in population
shifted_sample = df.sample(200, random_state=123).copy()
shifted_sample['age'] = shifted_sample['age'] * 1.3
shifted_sample['resting_bp'] = shifted_sample['resting_bp'] * 1.25

drift_report = analyze_dataset_drift(df, shifted_sample)
print(f"Pipeline Drift Status: {drift_report['overall_status']}")
pd.DataFrame(drift_report['feature_metrics'])
```
